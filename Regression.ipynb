```python
print("Sales unique values:")
print(df["sales"].unique())
print("\nVolume unique values:")
print(df["volume"].unique())
```

```python
for col in ["sales","volume"]:
    converted = pd.to_numeric(df[col],errors="coerce")
    invalid=df[df[col].notna()& converted.isna()]
    print(f"\nInvalid values in {col}:")
    print(invalid[col].unique())
```

```python
df["sales"]=pd.to_numeric(df["sales"],errors="coerce")
df["volume"]=pd.to_numeric(df["volume"],errors="coerce")
```

```python
print(df[["sales", "volume"]].isnull().sum())
```
